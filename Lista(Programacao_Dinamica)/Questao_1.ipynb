function Foo(A, n)

        return 0
    return Foo(A, ⌊n/2⌋) + Foo(A, ⌊n/2⌋ + 1) + Foo(A, ⌊n/2⌋ + 2) + A[n]
em que A é um vetor com posições de 1 até n.

Justificando sua resposta, diga se o algoritmo acima possui sobreposição de subproblemas. Em caso positivo, mostre um algoritmo com memoização para evitar recalcular subproblemas.

### Justificativa

Sim, o algoritmo possui sobreposição de subproblemas. Um subproblema é definido pela chamada da função `Foo` com um valor específico de `n`. Ao analisar a árvore de recursão, podemos observar que a mesma chamada de função com o mesmo argumento `n` ocorre múltiplas vezes.

Por exemplo, para uma chamada `Foo(A, 10)`, teremos as seguintes chamadas recursivas:
- `Foo(A, 10)` chama `Foo(A, 5)`, `Foo(A, 6)` e `Foo(A, 7)`.
- `Foo(A, 6)` chama `Foo(A, 3)`, `Foo(A, 4)` e `Foo(A, 5)`.
- `Foo(A, 7)` chama `Foo(A, 3)`, `Foo(A, 4)` e `Foo(A, 5)`.

Neste exemplo, o subproblema `Foo(A, 5)` é calculado três vezes, e os subproblemas `Foo(A, 3)` e `Foo(A, 4)` são calculados duas vezes cada. Essa repetição de cálculos para os mesmos subproblemas caracteriza a sobreposição.

Para evitar o recálculo, podemos usar a técnica de memoização, que armazena os resultados de subproblemas já resolvidos. Abaixo está a versão do algoritmo com memoização.


```python
def foo_memo(A, n, memo=None):
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]

    if 1 <= n <= 4:
        memo[n] = 0
    else:
        memo[n] = (
            foo_memo(A, n // 2, memo) +
            foo_memo(A, n // 2 + 1, memo) +
            foo_memo(A, n // 2 + 2, memo) +
            A[n]
        )

    return memo[n]
```
