# Questão 7
Você é o gerente de um projeto de desenvolvimento de software da empresa DP Health Solutions. O projeto é dividido em n fases, como planejamento, análise das necessidades do cliente, definição da arquitetura do sistema, definição das interfaces, definição dos fluxos de dados, codificação do software, testes de unidade, testes de integração, documentação, implantação e avaliação da implantação.

Você fez o orçamento de cada fase com três software houses. Todas elas conseguem realizar todas as fases do projeto. Para cada fase do projeto, cada software house define um orçamento potencialmente diferente das outras para concluir a fase. Além disso, uma software house pode concluir a primeira fase do projeto e depois outra software house pode realizar a segunda fase, e assim por diante. Dessa forma, também são estimados custos associados ao redirecionamento do projeto em cada fase entre as software houses. Por exemplo, temos um custo de redirecionamento da primeira software house para a segunda após a conclusão da fase de planejamento, temos um custo de redirecionamento da terceira software house para a primeira após a fase de definição das interfaces, e assim por diante.

Seu objetivo é determinar o menor custo necessário para completar todas as fases do projeto, levando em consideração os tempos de conclusão de cada fase em cada software house e os custos de redirecionamento entre elas.

Dessa forma, você tem n fases, e cada fase i tem um custo cost[i][j] para a j-ésima software house. Além disso, você tem um custo de redirecionamento redirect[i][j][k] para redirecionar o projeto da j-ésima software house para a k-ésima software house após a conclusão da fase i e para iniciar a fase i + 1. Usando programação dinâmica, você deve criar um algoritmo que, a partir da entrada n, cost e redirect, retorne o menor custo total para completar o projeto e qual software house ficará responsável por cada fase do projeto. Seu algoritmo deve resolver o problema em tempo O(n).

```python
def min_project_cost(n, cost, redirect):
    m = len(cost[0])  # número de software houses (assumido constante, por exemplo, 3)
    dp = [[float('inf')] * m for _ in range(n)]
    path = [[-1] * m for _ in range(n)]

    # Fase 0: só há o custo de cada software house fazer a primeira fase
    for j in range(m):
        dp[0][j] = cost[0][j]

    # Preenche DP para as demais fases
    for i in range(1, n):
        for curr in range(m):  # software house atual na fase i
            for prev in range(m):  # software house anterior na fase i-1
                redir_cost = redirect[i - 1][prev][curr]
                total_cost = dp[i - 1][prev] + redir_cost + cost[i][curr]
                if total_cost < dp[i][curr]:
                    dp[i][curr] = total_cost
                    path[i][curr] = prev  # salva a software house anterior

    # Encontra a última software house com menor custo
    min_cost = min(dp[n - 1])
    last_sh = dp[n - 1].index(min_cost)

    # Reconstrói o caminho de software houses
    schedule = [0] * n
    schedule[n - 1] = last_sh
    for i in range(n - 1, 0, -1):
        schedule[i - 1] = path[i][schedule[i]]

    return min_cost, schedule
```

```python
n = 4  # 4 fases
cost = [
    [5, 6, 4],   # fase 0
    [8, 5, 7],   # fase 1
    [6, 7, 5],   # fase 2
    [3, 6, 4],   # fase 3
]

redirect = [
    [  # após fase 0
        [0, 2, 3],
        [1, 0, 4],
        [2, 3, 0],
    ],
    [  # após fase 1
        [0, 2, 3],
        [2, 0, 1],
        [3, 1, 0],
    ],
    [  # após fase 2
        [0, 1, 2],
        [1, 0, 3],
        [2, 3, 0],
    ],
]

min_cost, schedule = min_project_cost(n, cost, redirect)
print("Menor custo:", min_cost)
print("Software house por fase:", schedule)
```
