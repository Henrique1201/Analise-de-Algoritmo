Modifique o algoritmo para resolver o problema das barras com **memoização** para ele retornar, além do rendimento ótimo, a **solução ótima**. Ou seja, sua solução deve mostrar (“printar”) os **tamanhos das barras da solução ótima**.

### Algoritmo original:

```python
function MEMOIZED-CUT-ROD(p, n)
    let r[0 : n] be a new array
    for i ← 0 to n do
        r[i] ← −∞
    end for
    return MEMOIZED-CUT-ROD-AUX(p, n, r)

function MEMOIZED-CUT-ROD-AUX(p, n, r)
    if r[n] ≥ 0 then ▷ Já tem solução para tamanho n?
        return r[n]
    if n = 1 then
        r[1] ← p[1]
        return p[1]
    r[n] ← p[n]
    for i ← 1 to n − 1 do
        r[n] ← max{r[n], p[i] + MEMOIZED-CUT-ROD-AUX(p, n − i, r)}
    end for
    return r[n]


````python
def memoized_cut_rod(p, n):
    r = [-float('inf')] * (n + 1)
    s = [0] * (n + 1)
    r[0] = 0

    def cut_rod_aux(n):
        if r[n] >= 0:
            return r[n]
        q = -float('inf')
        for i in range(1, n + 1):
            if i < len(p):
                temp = p[i] + cut_rod_aux(n - i)
                if temp > q:
                    q = temp
                    s[n] = i
        r[n] = q
        return q

    max_revenue = cut_rod_aux(n)

    print("Cortes ótimos:")
    while n > 0:
        print(f"Tamanho: {s[n]}")
        n -= s[n]

    return max_revenue
````

````python
p = [0, 1, 5, 8, 9, 10, 17, 17, 20, 24, 30]  
n = 7
print("Rendimento máximo:", memoized_cut_rod(p, n))
````
