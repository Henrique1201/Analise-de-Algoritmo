# Questão 3
Considere uma modificação do problema do corte de barras na qual, além do preço \( p_i \) de cada barra, cada corte incide um custo fixo \( c \). O rendimento da empresa associado com uma solução agora é a soma dos preços dos pedaços menos os custos dos cortes realizados.  
Usando programação dinâmica, apresente um algoritmo para resolver esse problema modificado. Observe que seu algoritmo deve retornar o rendimento da empresa e os tamanhos das barras cortados na solução ótima. Seu algoritmo deve ter tempo de execução de \( O(n^2) \).


```python
def cut_rod_with_cut_cost(p, n, c):
    r = [float('-inf')] * (n + 1)  # r[i]: lucro máximo para barra de tamanho i
    s = [0] * (n + 1)              # s[i]: primeiro corte ótimo para barra i
    r[0] = 0  # lucro com barra de tamanho 0 é 0

    for j in range(1, n + 1):  # para cada tamanho de barra
        max_revenue = p[j]  # caso não corte (venda inteira)
        s[j] = j  # assume que não cortamos
        for i in range(1, j):  # tenta todos os cortes possíveis
            revenue = p[i] + r[j - i] - c  # lucro fazendo corte
            if revenue > max_revenue:
                max_revenue = revenue
                s[j] = i  # salva melhor primeiro corte
        r[j] = max_revenue

    # reconstruindo os cortes ótimos
    cortes = []
    while n > 0:
        cortes.append(s[n])
        n -= s[n]

    return r[len(r) - 1], cortes
```

```python
p = [0, 1, 5, 8, 9, 10, 17, 17]  # preços p[i] para i = 1 a 7
n = 7  # tamanho da barra
c = 2  # custo por corte

lucro, cortes = cut_rod_with_cut_cost(p, n, c)

print("Lucro máximo:", lucro)
print("Cortes feitos:", cortes)
```
