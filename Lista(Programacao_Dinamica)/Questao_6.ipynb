# Questão 6
Você trabalha em uma fintech e ficou responsável pelo desenvolvimento de um módulo para avaliar o desempenho de carteiras de investimentos ao longo do tempo. A rentabilidade da carteira é registrada diariamente e pode variar, resultando em ganhos (valores positivos) ou perdas (valores negativos). Seu objetivo é identificar o período contínuo de tempo em que a carteira teve o melhor desempenho acumulado.

Dessa forma, dado um vetor de inteiros R onde R[i] representa a rentabilidade diária da carteira no dia i, encontre o subvetor contínuo (contendo pelo menos um número) que possui a maior soma e retorne essa soma.

**Exemplos**  
Para entrada R = [-2, 1, -3, 4, -1, 2, 1, -5, 4], temos 6 como soma máxima e a solução ótima é o subvetor [4, -1, 2, 1].  
Para entrada R = [5, 4, -1, 7, 8], temos 23 como soma máxima e a solução ótima é o subvetor [5, 4, -1, 7, 8], ou seja, o próprio vetor de entrada.

Usando programação dinâmica, apresente um algoritmo para resolver esse problema. Observe que seu algoritmo deve retornar a soma máxima e o subvetor com essa soma. Seu algoritmo deve executar em tempo O(n).


```python
def max_subvetor_soma(R):
    n = len(R)
    if n == 0:
        return 0, []

    # Inicializa as variáveis
    max_so_far = R[0]            # melhor soma encontrada até agora
    max_ending_here = R[0]       # soma que termina no índice atual

    start = 0    # início do subvetor ótimo
    end = 0      # fim do subvetor ótimo
    s = 0        # possível novo início

    for i in range(1, n):
        # Se for melhor recomeçar do elemento atual, faz isso
        if R[i] > max_ending_here + R[i]:
            max_ending_here = R[i]
            s = i
        else:
            max_ending_here += R[i]

        # Atualiza o melhor resultado
        if max_ending_here > max_so_far:
            max_so_far = max_ending_here
            start = s
            end = i

    # subvetor ótimo é de start a end (inclusive)
    return max_so_far, R[start:end+1]
```

```python
R1 = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
soma1, subvetor1 = max_subvetor_soma(R1)
print("Soma máxima:", soma1)
print("Subvetor ótimo:", subvetor1)
# Esperado: soma = 6, subvetor = [4, -1, 2, 1]

R2 = [5, 4, -1, 7, 8]
soma2, subvetor2 = max_subvetor_soma(R2)
print("Soma máxima:", soma2)
print("Subvetor ótimo:", subvetor2)
# Esperado: soma = 23, subvetor = [5, 4, -1, 7, 8]
```
