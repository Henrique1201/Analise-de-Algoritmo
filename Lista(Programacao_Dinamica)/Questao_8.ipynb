```markdown
Suponha que você receba uma imagem colorida representada por uma matriz `A[1 : m, 1 : n]` de tamanho `m×n`, onde cada elemento `A[i, j]` representa um pixel da imagem com uma tripla de intensidades de vermelho, verde e azul (RGB), respectivamente. Você deseja comprimir ligeiramente essa imagem, removendo um pixel de cada uma das `m` linhas, de modo que a imagem inteira fique uma coluna mais estreita. No entanto, para evitar efeitos visuais incongruentes, os pixels removidos em duas linhas adjacentes devem estar na mesma coluna ou em colunas adjacentes. Dessa forma, os pixels removidos formam uma costura (ou seam) da primeira até a última linha, onde pixels sucessivos na costura são adjacentes vertical ou diagonalmente.

Agora, suponha que junto com cada pixel `A[i, j]`, seja fornecida uma medida de perturbação `d[i, j] ∈ R`, indicando quão perturbador seria remover o pixel `A[i, j]`.

Intuitivamente, quanto menor a medida de perturbação de um pixel, mais semelhante ele é a seus vizinhos. Uma forma comum de calcular essa matriz de perturbação é por meio das diferenças de cor entre um pixel e seus vizinhos horizontais e verticais. Supondo que `A[i, j] = (Ri,j , Gi,j , Bi,j )`, uma definição possível para `d[i, j]` é:

```
d[i, j] = ∥A[i, j − 1] − A[i, j + 1]∥² + ∥A[i − 1, j] − A[i + 1, j]∥²
```

onde a distância entre dois pixels no espaço RGB é dada por:

```
∥A[p, q] − A[r, s]∥² = (Rp,q − Rr,s)² + (Gp,q − Gr,s)² + (Bp,q − Br,s)²
```

Esse cálculo mede a variação de cor nas direções horizontal e vertical. Em regiões homogêneas, onde os pixels são semelhantes aos vizinhos, o valor de `d[i, j]` será pequeno; em regiões com bordas ou detalhes visuais importantes, `d[i, j]` será maior.

A medida de perturbação de uma costura é definida como a soma das medidas de perturbação dos seus pixels. Projete um algoritmo que encontre uma costura com a menor medida de perturbação possível. Seu algoritmo deve ter complexidade de tempo `O(mn)`, onde `m` e `n` são, respectivamente, o número de linhas e colunas da imagem.
```

````python
def encontrar_costura_minima(d):
    m = len(d)        # número de linhas
    n = len(d[0])     # número de colunas

    # Matriz M para guardar menor custo até cada pixel na linha i
    M = [[0] * n for _ in range(m)]

    # Matriz para armazenar o caminho da costura (coluna anterior)
    caminho = [[0] * n for _ in range(m)]

    # Inicializa a primeira linha
    for j in range(n):
        M[0][j] = d[0][j]
        caminho[0][j] = -1  # não há linha anterior

    # Preenche a matriz M para linhas subsequentes
    for i in range(1, m):
        for j in range(n):
            # Colunas possíveis na linha anterior
            vizinhos = []
            for k in [j-1, j, j+1]:
                if 0 <= k < n:
                    vizinhos.append((M[i-1][k], k))
            min_custo, min_col = min(vizinhos, key=lambda x: x[0])

            M[i][j] = d[i][j] + min_custo
            caminho[i][j] = min_col

    # Encontra a coluna da última linha com menor custo
    min_total = float('inf')
    min_index = -1
    for j in range(n):
        if M[m-1][j] < min_total:
            min_total = M[m-1][j]
            min_index = j

    # Reconstrói a costura a partir da última linha
    costura = [0] * m
    costura[m-1] = min_index
    for i in range(m-1, 0, -1):
        costura[i-1] = caminho[i][costura[i]]

    return min_total, costura
````
