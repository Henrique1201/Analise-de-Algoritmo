# Questão 5
Você está desenvolvendo um aplicativo para planejar uma viagem de carro por uma série de cidades, onde cada cidade tem um custo associado, como alimentação, hospedagem e passeios turísticos. A entrada é um vetor de inteiros positivos `C` onde `C[i]` é o custo associado à parada na cidade `i`. Após pagar o custo de uma cidade, você pode escolher dirigir para a próxima cidade ou pular uma cidade, indo diretamente para a seguinte. Você pode iniciar sua viagem a partir da primeira ou da segunda cidade. Dessa forma, dado o vetor `C`, seu algoritmo deve retornar o custo mínimo para completar a viagem e alcançar o destino final.

**Exemplos**  
Para entrada `C = [10, 15, 20]`, temos `15` como custo mínimo. Uma solução ótima consiste de iniciar na segunda cidade e pular a terceira cidade.  
Para entrada `C = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]`, temos `6` como custo mínimo. Uma solução ótima consiste de iniciar na primeira cidade, depois ir para a terceira, depois para quinta, depois para a sétima, depois para a oitava, depois para a última.

Usando programação dinâmica, apresente um algoritmo para resolver esse problema. Observe que seu algoritmo deve retornar o custo mínimo da viagem e as cidades que serão visitadas. Seu algoritmo deve executar em tempo `O(n)`.


```python
def menor_custo_viagem(C):
    n = len(C)
    if n == 0:
        return 0, []
    if n == 1:
        return C[0], [0]

    # dp[i] = custo mínimo para chegar até a cidade i
    dp = [0] * n
    prev = [-1] * n  # para reconstruir o caminho

    # Podemos começar na cidade 0 ou 1
    dp[0] = C[0]
    dp[1] = C[1]

    # Nenhuma cidade anterior válida para as primeiras
    prev[0] = -1
    prev[1] = -1

    # Preenchemos o dp e prev com as melhores decisões
    for i in range(2, n):
        # Se viemos da cidade i-1
        custo1 = dp[i - 1] + C[i]
        # Se viemos pulando da cidade i-2
        custo2 = dp[i - 2] + C[i]

        # Escolhemos a opção com menor custo
        if custo1 < custo2:
            dp[i] = custo1
            prev[i] = i - 1
        else:
            dp[i] = custo2
            prev[i] = i - 2

    # Podemos terminar na última cidade (n-1) ou pulando e parando na penúltima (n-2)
    if dp[n - 1] < dp[n - 2]:
        custo_final = dp[n - 1]
        ultima_cidade = n - 1
    else:
        custo_final = dp[n - 2]
        ultima_cidade = n - 2

    # Reconstruir o caminho de trás pra frente
    caminho = []
    i = ultima_cidade
    while i >= 0:
        caminho.append(i)
        i = prev[i]
    caminho.reverse()  # agora temos em ordem da viagem

    return custo_final, caminho
```

```python
C1 = [10, 15, 20]
custo1, caminho1 = menor_custo_viagem(C1)
print("Custo mínimo:", custo1)
print("Cidades visitadas:", caminho1)
# Esperado: custo = 15, cidades = [1]

C2 = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]
custo2, caminho2 = menor_custo_viagem(C2)
print("Custo mínimo:", custo2)
print("Cidades visitadas:", caminho2)
# Esperado: custo = 6, cidades = [0, 2, 4, 6, 7, 9]
```
